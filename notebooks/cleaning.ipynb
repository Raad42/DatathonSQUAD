```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
train_path = os.path.join('..','data', 'raw', 'train.csv')
test_path = os.path.join('..', 'data', 'raw', 'test.csv')

trainDF = pd.read_csv(train_path)
testDF = pd.read_csv(test_path)

mainDF = pd.concat([trainDF, testDF], axis=0, ignore_index=True)
```

# Encoding Categorical Data

```python
#Mapping 
categorical_map = {
    'a' : 0,
    'b' : 1, 
    'c' : 2, 
    'd' : 3,
    'e' : 4, 
    'f' : 5, 
    'g' : 6, 
    'h' : 7, 
    'i' : 8,
    'j' : 9,
    'k' : 10,
    'l' : 11,
    'm' : 12,
    'n' : 13,
}
```

```python
# Apply mapping to trainDF
trainDF['X2'] = trainDF['X2'].map(categorical_map)
trainDF['X4'] = trainDF['X4'].map(categorical_map)
trainDF['X5'] = trainDF['X5'].map(categorical_map)
trainDF['X6'] = trainDF['X6'].map(categorical_map)
trainDF['X8'] = trainDF['X8'].map(categorical_map)
trainDF['X10'] = trainDF['X10'].map(categorical_map)
trainDF['X17'] = trainDF['X17'].map(categorical_map)
trainDF['X18'] = trainDF['X18'].map(categorical_map)

# Apply mapping to testDF
testDF['X2'] = testDF['X2'].map(categorical_map)
testDF['X4'] = testDF['X4'].map(categorical_map)
testDF['X5'] = testDF['X5'].map(categorical_map)
testDF['X6'] = testDF['X6'].map(categorical_map)
testDF['X8'] = testDF['X8'].map(categorical_map)
testDF['X10'] = testDF['X10'].map(categorical_map)
testDF['X17'] = testDF['X17'].map(categorical_map)
testDF['X18'] = testDF['X18'].map(categorical_map)
```

```python
# Split X9 into two columns based on month and year for trainDF
trainDF['X9_Month'] = trainDF['X9'].str.split('-').str[0]
trainDF['X9_Year'] = trainDF['X9'].str.split('-').str[1]
trainDF.drop(columns=['X9'], inplace=True)

# Split X9 into two columns based on month and year for testDF
testDF['X9_Month'] = testDF['X9'].str.split('-').str[0]
testDF['X9_Year'] = testDF['X9'].str.split('-').str[1]
testDF.drop(columns=['X9'], inplace=True)
```

```python
month_map = {
    'Jan' : 1,
    'Feb' : 2,
    'Mar' : 3,
    'Apr' : 4,
    'May' : 5,
    'Jun' : 6,
    'Jul' : 7,
    'Aug' : 8,
    'Sep' : 9,
    'Oct' : 10,
    'Nov' : 11,
    'Dec' : 12
}
```

```python
trainDF['X9_Month'] = trainDF['X9_Month'].map(month_map)
testDF['X9_Month'] = testDF['X9_Month'].map(month_map)

trainDF['X9_Month'], testDF['X9_Month']
```

# Missing Values

```python
trainDF['X15'] = trainDF['X15'].fillna(trainDF['X15'].mean())
trainDF['X19'] = trainDF['X19'].fillna(trainDF['X19'].mean())

testDF['X15'] = testDF['X15'].fillna(testDF['X15'].mean())
testDF['X19'] = testDF['X19'].fillna(testDF['X19'].mean())
```

```python
# Save the cleaned train and test data
trainDF.to_csv(os.path.join('..', 'data', 'processed', 'cleaned_train_data.csv'), index=False)
testDF.to_csv(os.path.join('..', 'data', 'processed', 'cleaned_test_data.csv'), index=False)
```

# Dealing with Skewed Data

```python
trainDF['X11'] = np.log1p(trainDF['X11'])
trainDF['X12'] = np.log1p(trainDF['X12'])
trainDF['X14'] = np.log1p(trainDF['X14'])

testDF['X11'] = np.log1p(testDF['X11'])
testDF['X12'] = np.log1p(testDF['X12'])
testDF['X14'] = np.log1p(testDF['X14'])
```

```python
# Saving the cleaned log-scaled data for train and test datasets
trainDF.to_csv(os.path.join('..', 'data', 'processed', 'cleaned_train_data_logScaled.csv'), index=False)
testDF.to_csv(os.path.join('..', 'data', 'processed', 'cleaned_test_data_logScaled.csv'), index=False)
```
