```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
train_path = os.path.join('..','data', 'raw', 'train.csv')
test_path = os.path.join('..', 'data', 'raw', 'test.csv')

trainDF = pd.read_csv(train_path)
testDF = pd.read_csv(test_path)

mainDF = pd.concat([trainDF, testDF], axis=0, ignore_index=True)
```

# Encoding Categorical Data

```python
#Mapping 
categorical_map = {
    'a' : 0,
    'b' : 1, 
    'c' : 2, 
    'd' : 3,
    'e' : 4, 
    'f' : 5, 
    'g' : 6, 
    'h' : 7, 
    'i' : 8,
    'j' : 9,
    'k' : 10,
    'l' : 11,
    'm' : 12,
    'n' : 13,
}
```

```python
mainDF['X2'] = mainDF['X2'].map(categorical_map)
mainDF['X4'] = mainDF['X4'].map(categorical_map)
mainDF['X5'] = mainDF['X5'].map(categorical_map)
mainDF['X6'] = mainDF['X6'].map(categorical_map)
mainDF['X8'] = mainDF['X8'].map(categorical_map)
mainDF['X10'] = mainDF['X10'].map(categorical_map)
mainDF['X17'] = mainDF['X17'].map(categorical_map)
mainDF['X18'] = mainDF['X18'].map(categorical_map)
```

```python
#Split X9 into two columns based on month and year
mainDF['X9_Month'] = mainDF['X9'].str.split('-').str[0]
mainDF['X9_Year'] = mainDF['X9'].str.split('-').str[1]
mainDF.drop(columns=['X9'], inplace=True)
```

```python
month_map = {
    'Jan' : 1,
    'Feb' : 2,
    'Mar' : 3,
    'Apr' : 4,
    'May' : 5,
    'Jun' : 6,
    'Jul' : 7,
    'Aug' : 8,
    'Sep' : 9,
    'Oct' : 10,
    'Nov' : 11,
    'Dec' : 12
}
```

```python
mainDF['X9_Month'] = mainDF['X9_Month'].map(month_map)
mainDF['X9_Month']
```

# Missing Values

```python
mainDF['X15'] = mainDF['X15'].fillna(mainDF['X15'].mean())
mainDF['X19'] = mainDF['X19'].fillna(mainDF['X19'].mean())
```

```python
#Saving the cleaned data
mainDF.to_csv(os.path.join('..', 'data', 'processed', 'cleaned_data.csv'), index=False)
```

# Dealing with Skewed Data

```python
mainDF['X11'] = np.log1p(mainDF['X11'])
mainDF['X12'] = np.log1p(mainDF['X12'])
mainDF['X14'] = np.log1p(mainDF['X14'])
```

```python
#Saving the cleaned log scaled data
mainDF.to_csv(os.path.join('..', 'data', 'processed', 'cleaned_data_logScaled.csv'), index=False)
```
