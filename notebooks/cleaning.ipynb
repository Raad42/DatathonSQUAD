```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
train_path = os.path.join('..','data', 'train.csv')
test_path = os.path.join('..', 'data', 'test.csv')

trainDF = pd.read_csv(train_path)
testDF = pd.read_csv(test_path)
```

**Encoding**
X2, X4

```python
#Mapping 
categorical_map = {
    'a' : 0,
    'b' : 1, 
    'c' : 2, 
    'd' : 3,
    'e' : 4, 
    'f' : 5, 
    'g' : 6, 
    'h' : 7, 
    'i' : 8,
    'j' : 9,
    'k' : 10,
    'l' : 11,
    'm' : 12,
    'n' : 13,
}
```

```python
trainDF['X2'] = trainDF['X2'].map(categorical_map)
trainDF['X4'] = trainDF['X4'].map(categorical_map)
trainDF['X5'] = trainDF['X5'].map(categorical_map)
trainDF['X6'] = trainDF['X6'].map(categorical_map)
trainDF['X8'] = trainDF['X8'].map(categorical_map)
trainDF['X10'] = trainDF['X10'].map(categorical_map)
```

# Dealing with Dates
