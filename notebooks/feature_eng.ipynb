# Correlation Analysis

```python
import pandas as pd
import os

filepath = os.path.join('..', 'data', 'processed', 'cleaned_train_data_logScaled.csv')

df = pd.read_csv(filepath)
```

```python
df.head(20)
```

```python
# Let's make and visualise a correlation matrix
correlation_matrix = df.corr()
correlation_matrix
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(20,16))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Correlation Matrix of Features")
plt.show()
```

```python
target = df['Y']
df = df.drop('Y', axis=1)
df = df.drop('ID', axis=1)
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_estimators=100, random_state=35)
rf.fit(df, target)
```

```python
importances = pd.Series(rf.feature_importances_, index=df.columns)
importances.sort_values(ascending=False).plot(kind='bar', figsize=(12,4), title="Random Forest Feature Importances")
plt.tight_layout()
plt.show()
```

```python
top_features = importances.sort_values(ascending=False).head(10).index
df_top = df[top_features]

df_top
```
