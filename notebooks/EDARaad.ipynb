```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
train_path = os.path.join('..','data', 'raw', 'train.csv')
test_path = os.path.join('..', 'data', 'raw', 'test.csv')

trainDF = pd.read_csv(train_path)
testDF = pd.read_csv(test_path)
```

```python
mainDF = pd.concat([trainDF, testDF], axis=1, ignore_index=False)
```

```python
mainDF.head()
```

**Y**

```python
# Plot frequency of Y = 0 or Y = 1
sns.countplot(x='Y', data=trainDF)
plt.title('Count of Y = 0 or Y = 1')        
plt.xlabel('Y')
plt.ylabel('Count')
plt.show()
```

**X1**

```python
trainDF['X1'].describe()
```

```python
trainDF['X1'].isna().sum()
```

```python
#Looking at feature X1
plt.figure(figsize=(10, 8))
sns.histplot(trainDF['X1'], bins=30, kde=True, color='blue')

plt.title('Histogram of X1', fontsize=16)
plt.xlabel('X1', fontsize=14)
plt.ylabel('Frequency', fontsize=14)

plt.show()
```

```python
log_transformed = np.log1p(trainDF['X1'])

plt.figure(figsize=(10, 8))
sns.histplot(log_transformed, bins=30, kde=True, color='blue')

plt.title('Log Transformed Histogram of X1', fontsize=16)
plt.xlabel('X1', fontsize=14)
plt.ylabel('Frequency', fontsize=14)
plt.show()
```

```python
trainTrueDF = trainDF[trainDF['Y'] != 0]

trainFalseDF = trainDF[trainDF['Y'] == 0]

fig, axs = plt.subplots(1, 2, figsize=(15, 5))
# Plots Boxplot for Data 1
axs[0].boxplot(trainTrueDF['X1'])
axs[0].set_title('Data 1')
axs[0].set_xlabel('Sample')
axs[0].set_ylabel('Value')
# Plots Boxplot for Data 2
axs[1].boxplot(trainFalseDF['X1'])
axs[1].set_title('Data 2')
axs[1].set_xlabel('Sample')
axs[1].set_ylabel('Value')

# Adjusts layout
plt.tight_layout()
plt.show()
```

```python


# Create figure and axes
fig, (ax1, ax2) = plt.subplots(nrows=1,  
                               ncols=2, 
                               figsize=(9, 4), 
                               sharey=True) 

# Plotting violin plot and boxplot for 'True'
ax1.set_title('True') 
ax1.set_ylabel('Observed values') 
ax1.violinplot(trainTrueDF['X1'], showmeans=True)
#ax1.boxplot(data1['X1'], widths=0.1, patch_artist=True, boxprops=dict(facecolor="white"))

# Plotting violin plot and boxplot for 'False'
ax2.set_title('False') 
ax2.violinplot(trainFalseDF['X1'], showmeans=True)
#ax2.boxplot(data2['X1'], widths=0.1, patch_artist=True, boxprops=dict(facecolor="white"))

# Show the plot
plt.tight_layout()
plt.show()
```
